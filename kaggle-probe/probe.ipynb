```python
import subprocess, sys, socket, os

print("=" * 60)
print("NETWORK")
print("=" * 60)
for host in ["pypi.org", "files.pythonhosted.org", "github.com", "kaggle.com"]:
    try:
        print(f"  DNS {host:<26} -> {socket.gethostbyname(host)}")
    except Exception as e:
        print(f"  DNS {host:<26} -> FAIL {type(e).__name__}: {e}")

for host, port in [("pypi.org", 443), ("files.pythonhosted.org", 443)]:
    try:
        s = socket.create_connection((host, port), timeout=8); s.close()
        print(f"  TCP {host}:{port} -> OK")
    except Exception as e:
        print(f"  TCP {host}:{port} -> FAIL {type(e).__name__}: {e}")

try:
    import urllib.request
    r = urllib.request.urlopen("https://pypi.org/simple/unsloth/", timeout=10)
    print(f"  HTTPS pypi.org/simple/unsloth/ -> {r.status}")
except Exception as e:
    print(f"  HTTPS pypi.org -> FAIL {type(e).__name__}: {e}")

print()
print("=" * 60)
print("PREINSTALLED (offline availability is what matters)")
print("=" * 60)
import importlib.metadata as md
for p in ["torch", "transformers", "trl", "peft", "accelerate", "bitsandbytes",
          "datasets", "unsloth", "gguf", "xformers", "sentencepiece", "protobuf",
          "huggingface-hub", "safetensors"]:
    try:
        print(f"  {p:<18} {md.version(p)}")
    except Exception:
        print(f"  {p:<18} NOT INSTALLED")

print()
print("=" * 60)
print("GPU + ENV")
print("=" * 60)
os.system("nvidia-smi")
try:
    import torch
    print("cuda:", torch.cuda.is_available(),
          torch.cuda.get_device_name(0) if torch.cuda.is_available() else "")
except Exception as e:
    print("torch import fail:", e)
print("KAGGLE_KERNEL_RUN_TYPE:", os.environ.get("KAGGLE_KERNEL_RUN_TYPE"))
print("proxy vars:", {k: v for k, v in os.environ.items() if "proxy" in k.lower()})
```
